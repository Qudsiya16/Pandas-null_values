#To check the null values in the data

```python
import pandas as pd
```

```python
data = {
    "Name" : ["Ali", "Sara", "Amjad", "Ayesha", "Noor"],
    "Age": [20, None, 21, None, 22],
    "English": [35, 43, None, None, 50],
    "Maths": [None, 45, 65, 23, None],
}
a=pd.DataFrame(data)
a
```

```python
#a.dropna() -->remove all the entries having null values.
#To check the null values in data
a.isnull().sum()
```

```python
#to fill the default values to make data clean and clear
b=a.fillna({"Age": 25, "English": 20, "Maths": 40})
b
```

```python
#Convert tabular data into CSV file.
b.to_csv("student.csv", index=False)
```

```python
#To read the file
c=pd.read_csv("student.csv")
c
```
